### Step 1. Import the necessary libraries

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv', sep='\t')
# Check that it worked
chipo.head()
```

### Step 4. How many products cost more than $10.00?

```python
# Is item_price a float? NO

chipo.item_price.dtype
```

```python
# Convert item_price to float

to_float = lambda x: float(x[1:]) if not isinstance(x, float) else x

chipo['item_price'] = chipo['item_price'].apply(to_float)

chipo.head()
```

```python
# Lets do some investigation

chipo.loc[chipo['item_name'] == 'Chicken Bowl']

# Item price seems to depend on exact choice_description and quantity
```

```python
# Add new column for price divided by quantity

chipo['individual_price'] = chipo['item_price'] / chipo['quantity']
chipo.head()
```

```python
# Merge on item_name and choice_description

merged = chipo.drop_duplicates(['item_name', 'choice_description'])

# Locate over $10

merged.loc[merged['individual_price'] > 10].shape[0]
```

### Step 5. What is the price of each item? 
###### print a data frame with only three columns item_name choice_description and product_price

```python
# Product price seems to be what we calculated as individual price:

chipo = chipo.rename(columns={'individual_price': 'product_price'})
merged = merged.rename(columns={'individual_price': 'product_price'})
```

```python
# From a peek at Solutions.ipynb, we only want when product_price > 10

dropped = merged[merged['product_price'] > 10]
dropped = dropped[['item_name', 'choice_description', 'product_price']]
dropped
```

### Step 6. Sort by the name of the item

```python
# Sort alphabetically
chipo_name_sorted = chipo.sort_values(by='item_name')
chipo_name_sorted
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
chipo.loc[chipo['item_price'].idxmax()]['quantity']
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
chipo.loc[chipo['item_name'] == 'Veggie Salad Bowl'].shape[0]
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
# Use loc with two conditions: named soda and more than one quantity
chipo.loc[
    (chipo['item_name'] == 'Canned Soda') & 
    (chipo['quantity'] > 1)
].shape[0]
```
